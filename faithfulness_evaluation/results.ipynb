```python
import os
import json
```

```python
count_graph =0
score_graph =0
```

```python
graph_output_dir = "path to output directory for CPG faithfulness"
profile_output_dir = "path to output directory for profile faithfulness"
```

```python
for filename in os.listdir(graph_output_dir):
    if filename.endswith(".json"):
        with open(graph_output_dir + "/" +filename) as f:
            d = json.load(f)
        score_graph = score_graph+d['score']
        count_graph = count_graph+1
print(score_graph/count_graph)
print(count_graph)
```
