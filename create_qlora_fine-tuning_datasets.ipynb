```python
import os
from os.path import isfile, join
from os import listdir
import json
import csv
from pathlib import Path
import glob
import random
import pandas as pd
```

```python
def generate_history(idx,history):
    history_idx = history[:idx]
    history_text = '\n'.join(
            [
                f"{message['speaker'].capitalize()}: {message['message']}"
                for message in history_idx
            ]
    )

    return history_text
```

```python
output_dir = Path("qlora_datasets/")
output_dir.mkdir(parents=True, exist_ok=True)
```

## Graph2Counsel

```python
final_dict = {}
```

```python
final_dict["data"] = []
```

```python
directory_path = "path to counseling dialogues files directory"
directory_path_profile = "path to profile files directory"
```

```python
files_list = [f for f in listdir(directory_path) if isfile(join(directory_path, f))]
```

```python
for file in files_list:
    with open(join(directory_path,file)) as f:
        data = json.load(f)
    filename_profile = Path(file).stem + ".txt"
    file_path_profile = join(directory_path_profile, filename_profile)
    with open(file_path_profile, 'r', encoding='utf-8') as f_profile:
        client_profile = f_profile.read()
    for x in range(2,len(data["History"]),2):
            final_dict["data"].append({"history":generate_history(x,data["History"]), "response":data["History"][x]["message"],"profile":client_profile})
```

```python
with open('qlora_datasets/graph2counsel-train.json', 'w') as fp:
    json.dump(final_dict, fp)
```

## MAGneT

```python
final_dict = {}
```

```python
final_dict["data"] = []
```

```python
for filename in os.listdir("path to counseling dialogues files directory"):
    if filename.endswith(".json"):
        with open("path to counseling dialogues files directory" + filename) as json_file:
                json_data = json.load(json_file)
        for x in range(2,len(json_data["history"]),2):
            final_dict["data"].append({"history":generate_history(x,json_data["history"]), "response":json_data["history"][x]["message"],"client_information":json_data['example']['AI_counselor']['CBT']['client_information'], "reason_counseling":json_data['example']['AI_counselor']['CBT']['reason_counseling']})
```

```python
with open('qlora_datasets/magnet-train.json', 'w') as fp:
    json.dump(final_dict, fp)
```
